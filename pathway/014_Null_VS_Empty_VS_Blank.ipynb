### DATA POLICIES

#### Set of rules that defines how data should be handled

1. Only use NULLs and empty strings, but avoid blank spaces (USE TRIM)

```SQL
WITH Orders AS (
    SELECT 1 Id, 'A' Category UNION
    SELECT 2, NULL UNION
    SELECT 3, '' UNION
    SELECT 4, ' '
)
SELECT
    *,
    LENGTH(TRIM(Category)) AS CategoryLen
FROM Orders;
```

````python
df2 = pd.DataFrame(data)

df2['categorylen'] = df2['Category'].str.strip().str.len()

df2
````

### 2 DATA POLICY

2. Only use NULLS and avoid using empty strings and blanck spaces

```SQL
WITH Orders AS (
    SELECT 1 Id, 'A' Category UNION
    SELECT 2, NULL UNION
    SELECT 3, '' UNION
    SELECT 4, ' '
)
SELECT
    *,
    NULLIF(TRIM(Category),'') AS null_empty
FROM Orders;
```

````python
df3 = pd.DataFrame(data)

df3['Category'] = df3['Category'].str.strip()

df3['Category'] = df3['Category'].replace('', np.nan)

df3['Category'] = df3['Category'].fillna(value=np.nan)

df3
````

### 3 DATA POLICY

3. Use the default value 'unknown' and avoid using nulls, empty strings, and blanck spaces

```SQL
WITH Orders AS (
    SELECT 1 Id, 'A' Category UNION
    SELECT 2, NULL UNION
    SELECT 3, '' UNION
    SELECT 4, ' '
)
SELECT
    COALESCE(NULLIF(TRIM(Category),''), 'unknown') AS Policy3
FROM Orders;
```
