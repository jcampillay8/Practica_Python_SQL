# 007 - Joins in SQL

````python
import pandas as pd
import numpy as np
````

````python
df_customers = pd.read_csv('data/customers.csv')
df_orders = pd.read_csv('data/orders.csv')

df_customers
````

````python
df_orders
````

# INNER JOIN

### Only Matching Data

![Inner Join](Pictures/Inner_Join.png)

### SQL TASK 1
#### Get all customers along with their orders, but only for customers who have placed an order

```SQL
SELECT
    *
FROM customers c
INNER JOIN orders o ON c.id = o.customer_id;
```

````python
df_result = pd.merge(
    df_customers,
    df_orders,
    left_on='id',
    right_on='customer_id',
    how='inner'
)

df_final = df_result[['id','first_name','order_id','sales']]

df_final
````

# LEFT JOIN

### Returns All rows from left and Only Matching From Right

![Left Join](Pictures/Left_Join.png)

### SQL TASK 2

#### Get all customers along with their orders, including those without orders

```SQL
SELECT
    c.id,
    c.first_name,
    o.order_id,
    o.sales
FROM customers c
LEFT JOIN orders o ON c.id = o.customer_id;
```

````python
df_result_left = pd.merge(
    df_customers,
    df_orders,
    left_on = 'id',
    right_on = 'customer_id',
    how = 'left'
)

df_final_left = df_result_left[['id','first_name','order_id','sales']]

df_final_left
````

### INNER JOIN

### Returns All Rows from Right and Only Matching From Left

![right Join](Pictures/Right_Join.png)

### SQL TASK 3

#### Get all customers along with their orders, including orders without matching customers

```SQL
SELECT
    c.id,
    c.first_name,
    o.order_id,
    o.sales
FROM customers c
RIGHT JOIN orders o ON c.id = o.customer_id;
```

````python
df_result_right = pd.merge(
    df_customers,
    df_orders,
    left_on = 'id',
    right_on = 'customer_id',
    how='right'
)

df_final_right = df_result_right[['id','first_name','order_id','sales']]

df_final_right
````

### FULL JOIN

#### Returns All Rows from Both Tables

![Full Join](Pictures/Full_Join.png)


### SQL TASK 4

#### Get all customers and all orders, even if there's no match

```SQL
SELECT
    c.id,
    c.first_name,
    o.order_id,
    o.sales
FROM customers c
FULL JOIN orders o ON c.id = o.customer_id;
```

````python
df_result_full = pd.merge(
    df_customers,
    df_orders,
    left_on = 'id',
    right_on = 'customer_id',
    how='outer'
)

df_final_full = df_result_full[['id','first_name','order_id','sales']]

df_final_full
````
