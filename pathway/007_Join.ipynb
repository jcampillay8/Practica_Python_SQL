# INNER JOIN

### Only Matching Data

![Inner Join](Pictures/Inner_Join.png)

### SQL TASK 1
#### Get all customers along with their orders, but only for customers who have placed an order

```SQL
SELECT
    *
FROM customers c
INNER JOIN orders o ON c.id = o.customer_id;
```

````python
df_result = pd.merge(
    df_customers,
    df_orders,
    left_on='id',
    right_on='customer_id',
    how='inner'
)

df_final = df_result[['id','first_name','order_id','sales']]

df_final
````

# LEFT JOIN

### Returns All rows from left and Only Matching From Right

![Left Join](Pictures/Left_Join.png)

### SQL TASK 2

#### Get all customers along with their orders, including those without orders

```SQL
SELECT
    c.id,
    c.first_name,
    o.order_id,
    o.sales
FROM customers c
LEFT JOIN orders o ON c.id = o.customer_id;
```

````python
df_result_left = pd.merge(
    df_customers,
    df_orders,
    left_on = 'id',
    right_on = 'customer_id',
    how = 'left'
)

df_final_left = df_result_left[['id','first_name','order_id','sales']]

df_final_left
````

### INNER JOIN

### Returns All Rows from Right and Only Matching From Left

![right Join](Pictures/Right_Join.png)

### SQL TASK 3

#### Get all customers along with their orders, including orders without matching customers

```SQL
SELECT
    c.id,

RIGHT JOIN orders o ON c.id = o.customer_id;
```

````python
df_result_right = pd.merge(
    df_customers,
    df_orders,
    left_on = 'id',
    right_on = 'customer_id',
    how='right'
)

df_final_right = df_result_right[['id','first_name','order_id','sales']]

df_final_right
````

### FULL JOIN

#### Returns All Rows from Both Tables

![Full Join](Pictures/Full_Join.png)


### SQL TASK 4

#### Get all customers and all orders, even if there's no match

```SQL
SELECT
    c.id,
    c.first_name,
    o.order_id,
    o.sales
FROM customers c
FULL JOIN orders o ON c.id = o.customer_id;
```

````python
df_result_full = pd.merge(
    df_customers,
    df_orders,
    left_on = 'id',
    right_on = 'customer_id',
    how='outer'
)

df_final_full = df_result_full[['id','first_name','order_id','sales']]

df_final_full
````

### LEFT ANTI JOIN

###  Returns Rows From Left that has No Match in Right

![Left Anti Join](Pictures/Left_Anti_Join.png)



### SQL TASK 5

#### Get all customers who haven't placed any order

```SQL
SELECT
    c.id,
    c.first_name,
    o.order_id,
    o.sales
FROM customers c
LEFT JOIN orders o ON c.id = o.customer_id
WHERE o.customer_id IS NULL;
```

````python
# 1. Realizar el LEFT JOIN
df_merged = pd.merge(
    df_customers[['id', 'first_name']], 
    df_orders[['customer_id', 'order_id', 'sales']], 
    left_on='id', 
    right_on='customer_id', 
    how='left'
)

# 2. Filtrar donde la columna de la derecha sea nula (Equivalente al WHERE IS NULL)
df_no_orders = df_merged[df_merged['customer_id'].isna()]

# 3. Seleccionar las columnas finales y limpiar
df_final = df_no_orders[['id', 'first_name', 'order_id', 'sales']]

df_final
````

### SQL TASK 6

#### Find customers without orders and orders without customers

```SQL
SELECT
    c.id,
    c.first_name,
    o.order_id,
    o.sales
FROM customers c
FULL JOIN orders o ON c.id = o.customer_id
WHERE c.id IS NULL OR o.customer_id IS NULL;
```

````python
# 1. Realizar el FULL OUTER JOIN con el indicador de origen
df_merged = pd.merge(
    df_customers[['id', 'first_name']], 
    df_orders[['customer_id', 'order_id', 'sales']], 
    left_on='id', 
    right_on='customer_id', 
    how='outer',
    indicator=True
)

# 2. Filtrar para mantener solo los registros que NO están en ambas tablas
# El indicador '_merge' tendrá el valor 'both' si hay coincidencia.
df_exclusive = df_merged[df_merged['_merge'] != 'both']

# 3. Seleccionar columnas finales
df_final = df_exclusive[['id', 'first_name', 'order_id', 'sales']]

df_final
````
