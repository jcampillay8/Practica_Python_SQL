# 📝 Reto Fase 1: Selección y Alias
Imagina que el Gerente de Ventas te pide un reporte rápido de los productos disponibles, pero lo quiere con nombres específicos para que sea más fácil de leer en una presentación.

La pregunta: Extrae de la tabla sales_products.csv las columnas product y price, pero cámbiales el nombre para que se vean como Nombre_Producto y Precio_Unitario respectivamente.

```SQL
SELECT
    product AS Nombre_Producto,
    price AS Precio_Unitario
FROM sales.products;
```

````python
# PANDAS
df_p = df_products.copy()

resultado = df_products[['product','price']].rename(
    columns={'product': 'Nombre_Producto', 'price':'Precio_Unitario'}
)

resultado
````

````python
pl_p = pl_products

resultado_pl = pl_p.select([
    pl.col('product').alias('Nombre_Producto'),
    pl.col('price').alias('Precio_Unitario')
])

resultado_pl
````

# 📝 Mini-Reto Final Fase 1 (Para cerrar con broche de oro)

De la tabla sales_customers.csv, selecciona firstname, lastname y score. Pero llámalos: Nombre, Apellido y Puntaje. 

```SQL
SELECT
    firstname AS Nombre,
    lastname AS Apellido,
    score AS Puntaje
FROM sales.customers;
```

````python
#Pandas

df_c = df_customers.copy()

respuesta = df_c[['firstname','lastname','score']].rename(
    columns={'firstname':'Nombre','lastname':'Apellido', 'score':'Puntaje'}
)

respuesta
````

````python
# Polars
pl_c = pl_customers

respuesta = pl_c.select([
    pl.col('firstname').alias('Nombre'),
    pl.col('lastname').alias('Apellido'),
    pl.col('score').alias('Puntaje')
])

respuesta
````

# 📝 El Reto Final de la Fase 1 (Combinado)
Vamos a juntar todo lo aprendido en la Fase 1. Usando la tabla sales_orders.csv, necesito que hagas lo siguiente:

Selecciona la columna orderstatus.

Asegúrate de que los valores no se repitan (queremos ver los estados únicos de las órdenes).

Cambia el nombre de la columna a Estado_Unico.

Muestra solo los primeros 2 resultados.

```SQL
SELECT
    DISTINCT orderstatus AS Estado_Unico
FROM sales.orders
LIMIT 2;
```

````python
# Pandas
df_o = df_orders.copy()

respuesta = df_o[['orderstatus']].drop_duplicates().rename(columns={'orderstatus':'EStadoUnico'}).head(2)

respuesta
````
