### CONCAT

```SQL
SELECT
    firstname,
    country,
    CONCAT (firstname,' ',country) AS name_country
FROM sales.customers;
```

````python
df_resultado = df_customers[['firstname','country']].copy()

df_resultado['name_country'] = df_resultado['firstname']+' '+df_resultado['country']

df_resultado
````

### UPPER and LOWER

```SQL
SELECT
    UPPER(firstname)
FROM sales.customers;

SELECT
    LOWER(firstname)
FROM sales.customers;
```

````python
df_customers['firstname_upper'] = df_customers['firstname'].str.upper()

df_customers['firstname_lower'] = df_customers['firstname'].str.lower()

df_customers
````

### TRIM

```SQL
SELECT
    TRIM(firstname)
FROM sales.customers;
```

````python
df_customers['firstname_trim'] = df_customers['firstname'].str.strip()

df_customers
````

### REPLACE 

```SQL
SELECT
    '123-456-7890' AS phone,
    REPLACE('123-456-7890','-','') AS clean_phone
```

````python
df = pd.DataFrame({'phone': ['123-456-7890','987-654-3210']})

df['clean_phone'] = df['phone'].str.replace( '-', '', regex=False)

df
````

### LEFT and RIGHT

```SQL
SELECT
    LEFT(firstname,2),
    RIGHT(firstname, 2)
FROM sales.customers;
```

````python
df_customers['first_2'] = df_customers['firstname'].str[:2]
df_customers['last_2'] = df_customers['firstname'].str[-2:]

df_customers
````

### SUBSTRING (VAlUE, START, LENGTH)

```SQL
SELECT
    SUBSTRING(firstname, 2, LENGTH(firstname)) AS Sub_name
FROM sales.customers;
```

````python
df_customers['Sub_Name'] = df_customers['firstname'].str[1:]

df_customers
````

### GETDATE() | CURRENT_TIMESTAMP

```SQL

```


### EXTRACT (YEAR FROM ...)

```SQL
SELECT
    orderid,
    creationtime,
    EXTRACT (YEAR FROM creationtime) AS order_year,
    EXTRACT (MONTH FROM creationtime) AS order_month,
    EXTRACT (DAY FROM creationtime) AS order_day,
    EXTRACT (HOUR FROM creationtime) AS order_hour,
    EXTRACT (MINUTE FROM creationtime) AS order_minute,
    EXTRACT (SECOND FROM creationtime) AS order_seconds
FROM sales.orders;
```

````python
# Aseguramos que sea formato fecha
df_orders['creationtime'] = pd.to_datetime(df_orders['creationtime'])

# Extraemos cada componente
df_orders['order_year'] = df_orders['creationtime'].dt.year
df_orders['order_month'] = df_orders['creationtime'].dt.month
df_orders['order_day'] = df_orders['creationtime'].dt.day
df_orders['order_hour'] = df_orders['creationtime'].dt.hour
df_orders['order_minute'] = df_orders['creationtime'].dt.minute
df_orders['order_seconds'] = df_orders['creationtime'].dt.second

df_orders
````

### date_part

```SQL
SELECT
    orderid,
    creationtime,
    date_part('quarter', creationtime) as quarter_dp,
    date_part('week', creationtime) as week_dp
FROM sales.orders;
```

````python
# Trimestre
df_orders['quarter_dp'] = df_orders['creationtime'].dt.quarter

# Semana (ISO)
df_orders['week_dp'] = df_orders['creationtime'].dt.isocalendar().week

df_orders
````
