# 006 - DML Commands (INSERT UPDATE DELETE)

````python
import pandas as pd
import numpy as np
````

````python
df_customer = pd.read_csv('data/customers.csv')

df_customer.head()
````

```SQL
INSERT INTO customers (id, first_name, country, score)
VALUES
    (6,'Anna','USA',NULL),
    (7,'Sam',NULL,100)
```

````python
nuevos_datos = pd.DataFrame([
    {"id": 6, "first_name": "Anna", 'country': "USA", 'score': None},
    {"id": 7, "first_name": "Sam", 'country': None, 'score': 100}
])

df_customer = pd.concat([df_customer, nuevos_datos], ignore_index=True)
````

````python
df_customer
````

````python
bad_data = pd.DataFrame([
    {'id':9,'first_name': 'USA', 'country':'Max','score':'200'}
])

df_customer = pd.concat([df_customer, bad_data], ignore_index=True)

````

````python
df_customer = df_customer[df_customer['id'] != 9]
````

````python
df_customer
````

# Copy data from 'customers' table into 'persons'

```SQL
INSERT INTO persons (id, person_name, birthday, phone)
SELECT
    id,
    first_name,
    NULL,
    'Unkown'
FROM customers
```

# Change the score of customer with ID 6 to 0

```SQL
UPDATE customers
SET score = 0
WHERE id BETWEEN 0 and 6;
```

````python
df_customer.loc[df_customer['id'].between(0,6), 'score'] = 0
````

````python
df_customer
````

```SQL
UPDATE customers
SET
    score = 0,
    country = 'UK'
WHERE id BETWEEN 0 and 6;
```

````python
df_customer.loc[df_customer['id'].between(0,6), ['score','country']] = [10,'UK']
````

````python
df_customer
````
