# 009 - Level9 

## 9. Performance Thinking

Optimizar consultas no es solo escribir SQL que funcione, sino SQL que **escale bien** y use los recursos de forma eficiente.

### 🚀 Conceptos Clave

- **Indexes**  
  Aceleran la ejecución de consultas en datasets grandes al permitir búsquedas más rápidas.

- **EXPLAIN / EXPLAIN ANALYZE**  
  Permiten entender **cómo** la base de datos ejecuta una consulta, mostrando costos, uso de índices y orden de operaciones.

- **Evitar `SELECT *`**  
  Reduce el uso de memoria y la cantidad de datos transferidos.  
  Selecciona solo las columnas que realmente necesitas.

- **Filtering Early**  
  Aplicar filtros lo antes posible (en `WHERE` o subconsultas) mejora el rendimiento de **Joins** y **Agregaciones**.
