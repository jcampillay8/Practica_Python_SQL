## 🔄 Equivalencias SQL vs Pandas vs Polars

| Operación SQL | Pandas 🐼                              | Polars 🧊                                      |
|---------------|----------------------------------------|-----------------------------------------------|
| `SELECT`      | `df[['col1', 'col2']]`                 | `.select(['col1', 'col2'])`                   |
| `DISTINCT`    | `.drop_duplicates()`                   | `.unique()`                                   |
| `WHERE`       | `df[df['col'] > 10]`                   | `.filter(pl.col('col') > 10)`                 |
| `GROUP BY`    | `.groupby('col').agg(...)`             | `.group_by('col').agg(...)`                   |
| `HAVING`      | `.groupby(...).filter(lambda x: x['col'].sum() > 5)` | `.group_by(...).agg(...).filter(pl.col('col_calculada') > 5)` |
| `ORDER BY`    | `.sort_values()`                       | `.sort()`                                     |
| `LIMIT`       | `.head(n)`                             | `.head(n)`                                    |
| `CASE WHEN`   | `np.where()` o `.map()`                | `pl.when().then().otherwise()`                |
