````python
import pandas as pd
import numpy as np
import polars as pl
from datetime import date
````

````python
df_ads = pd.read_csv('../Data/004/ad_campaigns.csv', parse_dates=['start_date'])
````

````python
pl_ads = pl.read_csv('../Data/004/ad_campaigns.csv', try_parse_dates=True)
````

# Pregunta 1

### Identifica los tipos de anuncios (ad_type) únicos que tuvieron más de 100 conversiones y que comenzaron en mayo de 2024.

```SQL
SELECT
    DISTINCT ad_type
FROM ad_campaigns
WHERE conversions > 100 AND
      start_date >= '2024-05-01';
```

````python
res = df_ads[
    (df_ads['conversions'] > 100) &
    (df_ads['start_date'] >= '2024-05-01')
]

res = res[['ad_type']].drop_duplicates().reset_index(drop=True)
````

````python
res = (pl_ads.filter(
    (pl.col('start_date') >= (date(2024,5,1))) &
    (pl.col('conversions') > 100)
).select('ad_type').unique()
)
````

# Pregunta 2

### De las campañas que comenzaron en junio de 2024, encuentra aquellas que no generaron ningún ingreso (revenue). Por favor, muestra los nombres de las campañas (campaign_name) y sus fechas de inicio (start_date).

```SQL
SELECT
    campaign_name,
    start_date
FROM ad_campaigns
WHERE (start_date BETWEEN '2024-06-01' AND '2024-06-30') AND
      revenue = 0;
```

````python
jun = df_ads[
    (df_ads['start_date'].between('2024-06-01','2024-06-30')) &
    (df_ads['revenue'] == 0)
]

res = jun[['campaign_name','start_date']]
````

````python
res = pl_ads.filter(
    (pl.col('start_date').is_between(date(2024,6,1), date(2024,6,30))) &
    (pl.col('revenue') == 0)
).select(
    ['campaign_name','start_date']
)

res
````
