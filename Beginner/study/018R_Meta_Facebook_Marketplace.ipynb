# Meta - Facebook Marketplace Item Price Analyses

````python
import pandas as pd
import numpy as np
import polars as pl
from datetime import date
````

````python
df_listing = pd.read_csv('../Data/018/listings.csv')

pl_listing = pl.read_csv('../Data/018/listings.csv')
````

# Pregunta 1

### ¿Puedes encontrar el precio promedio de los artículos listados en cada categoría en Facebook Marketplace? Queremos entender las tendencias de precios en las diferentes categorías.

```SQL
SELECT
    category,
    AVG(price)
FROM Listings
GROUP BY category;
```

````python
res = df_listing.groupby('category').agg(
    precio_promedio = ('price', 'mean')
).reset_index()
````

````python
res = pl_listing.group_by('category').agg(
    pl.col('price').mean().alias('precio_promedio')
)
````

# Pregunta 2

### ¿Qué ciudad tiene el precio promedio más bajo? Esto nos ayudará a identificar las ciudades más asequibles (baratas) para los compradores.

```SQL
SELECT
    city,
    AVG(price) AS avg_price_city
FROM Listings
GROUP BY city
ORDER BY  avg_price_city ASC
```

````python
res = df_listing.groupby('city').agg(
    avg_price_city = ('price', 'mean')
).sort_values(by='avg_price_city', ascending=True).head(1)
````

````python
res = pl_listing.group_by('city').agg(
    pl.col('price').mean().alias('avg_price_city')
).sort('avg_price_city', descending=False).head(1)

res
````
