# Pregunta 1

### Airbnb desea destacar los alojamientos con mejor calificación en su página de inicio. Escriba una consulta que devuelva el nombre (name), el barrio (neighborhood) y la calificación (rating) de todos los alojamientos que tengan una calificación de 4.5 o superior.

```SQL
SELECT
    name,
    neighborhood,
    rating
FROM listings
WHERE rating >= 4.5
```

````python
res = df_listing[
    df_listing['rating'] >= 4.5
].reset_index()

res = res[['name','neighborhood','rating']]
````

````python
res = pl_listing.filter(
    pl.col('rating') >= 4.5
).select(
    pl.col('name'),
    pl.col('neighborhood'),
    pl.col('rating')
)
````
